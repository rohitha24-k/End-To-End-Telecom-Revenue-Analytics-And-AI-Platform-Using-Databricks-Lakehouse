```python
import pyspark.sql.functions as F
```

```python
%sql
USE CATALOG liquid_telecom;
```

## **Account**

```python
raw_data_path = f"/Volumes/liquid_telecom/raw/liquid_telecom_volume/dim/accounts"

df = spark.read.option('header', "true").option("delimeter", ",").csv(raw_data_path)

# add metadata columns
df = df.withColumn("_source_file", F.col("_metadata.file_path")) \
       .withColumn("ingested_at", F.current_timestamp())
       
df.write.format("delta") \
    .mode("overwrite") \
    .option("mergeSchema", "true") \
    .saveAsTable("liquid_telecom.bronze.b_dim_account")
```
